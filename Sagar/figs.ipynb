```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, chi2
```

```python
xx = np.linspace(0, 10, 1000)
plt.plot(xx, norm.pdf(xx, 5, 1), color='black', linewidth=5)
```

```python
xx2 = np.linspace(0, 15, 1000)
plt.plot(xx2, chi2.pdf(xx2, df=3), color='black', linewidth=5)
```
